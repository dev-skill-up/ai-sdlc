```{=latex}
\begin{frame}[fragile]
\frametitle{Two Narratives}

``AI tanked our code quality'' \pause

``AI tripled our productivity'' \pause

Both are right.

\end{frame}
```

There are two loud camps in the AI-assisted coding debate. One says quality cratered. The other says productivity exploded. Both have data. Both are vocal. My claim is that they are both right, and they are experiencing the same underlying effect from opposite sides.

```{=latex}
\begin{frame}[fragile]
\frametitle{Guardrails Compound}

Automation catches recurring mistakes \pause

Gentler, cheaper, scales effortlessly \pause

Without them, entropy is directional

\end{frame}
```

Guardrails, linters, tests, CI checks, catch recurring mistakes before a human ever sees them. They are gentler than review comments, cheaper than context switches, and they scale. Teams that compound guardrails over years accumulate cognitive capacity. Teams that do not watch entropy grow: humans get tired, violations become precedent, inconsistency wins.

```{=latex}
\begin{frame}[fragile]
\frametitle{AI Changes the Velocity}

Productivity: 3--4x, and accelerating \pause

Each model release widens the gap \pause

Guardrails are now load-bearing

\end{frame}
```

AI has dramatically increased developer productivity: 3 to 4 times, and climbing with each model release. That is the real issue. Weak guardrails plus low velocity was barely functional. Weak guardrails plus AI velocity is a quality crisis. Strong guardrails plus AI velocity just ships more features. The velocity change is what makes process quality load-bearing instead of optional.

```{=latex}
\begin{frame}[fragile]
\frametitle{The Organizational Trap}

Easy response: blame the AI \pause

Correct response: fix the processes \pause

The maturity to fix it is the maturity that prevents it

\end{frame}
```

Blaming the AI is politically costless. Nobody has to admit past practices were inadequate. Fixing the processes requires cultural change and infrastructure investment. Here is the trap: the organizational maturity needed to respond correctly is the same maturity that would have prevented the problem. The teams struggling most are, by definition, the teams least equipped to fix it, and the gap is widening at an accelerating rate.

```{=latex}
\begin{frame}[fragile]
\frametitle{What To Do}

If AI exposed quality problems, you had quality problems \pause

Audit what review catches; automate it \pause

Make correctness the path of least resistance

\end{frame}
```

If AI exposed quality problems, you had quality problems. AI just raised velocity until they became visible. The concrete action: audit what your review process currently catches, and invest in automation proportional to the velocity gains you are seeing. Make correctness the path of least resistance, not a tax paid on every change.

```{=latex}
\begin{frame}[fragile]
\frametitle{This Week}

Pick one comment you leave in code reviews repeatedly \pause

Turn it into a linter rule, test, or CI check \pause

Ship it before your next AI-assisted PR

\vspace{1em}
\small Moshe Zadka -- \url{https://cobordism.com}

\end{frame}
```
